```python
import pandas as pd
import numpy as np
```

## 1.Create datafram from an array

### 1.1 Option 1

```python
# Creating an array
data = np.array([[1,4], [2,5], [3,6]])
```

```python
# Creating a dataframe
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'], columns=['col1', 'col2'])
```

```python
# Showing the dataframe
df
```

### 1.2 Option 2


```python
# Creating an array with list shape
data = [[1,4], [2,5], [3,6]]
```

```python
# Creating a dataframe
df = pd.DataFrame(data, index=['row1', 'row2', 'row3'], columns=['col1', 'col2'])
```

```python
# Showing the dataframe
df
```

### 1.3 Option 3


```python
pd.DataFrame(data={'Bob': ['I liked it.', 'It was awful.'], 
              'Sue': ['Pretty good.', 'Bland.']},
             index=['Product A', 'Product B'])
```

## 2. Create a Dataframe from dictionary

```python
# list used for this example
states = ['California', 'Texas', 'Florida', 'New York']
population = [39538223, 29145505, 21538187, 20201249]
```

```python
# Storing list within a dictionary
dict_states = {'States' : states, 'Population': population}
```

```python
# Creating the dataframe
df_population = pd.DataFrame(dict_states)
```

```python
# Showing the dataframe
df_population
```
