```python
import pandas as pd
```

```python
df = pd.read_csv('../titanic3.csv')
df.head()
```

```python
# describe():  tạo ra một bản tóm tắt cấp cao về các thuộc tính của cột được chỉ định. Phương pháp này nhận biết kiểu dữ liệu, 
#               nghĩa là đầu ra của nó thay đổi dựa trên kiểu dữ liệu của đầu vào.
df.age.describe() # Đối với dữ liệu số
```

```python
df['sex'].describe() # Đối với dữ liệu chuỗi
```

```python
# mean() : số trung bình
df.age.mean()
```

```python
# unique() : danh sách các giá trị duy nhất
df.ticket.unique()
```
