__Map__


- một hàm lấy một tập giá trị và "ánh xạ" chúng thành một tập giá trị khác.

- Trong khoa học dữ liệu, tạo ra các biểu diễn mới từ dữ liệu hiện có, hoặc chuyển đổi dữ liệu từ định dạng hiện tại sang định dạng mong muốn sau này

```python
import pandas as pd
```

```python
df = pd.read_csv('../titanic3.csv')
df.head()
```

```python
df_age_mean = df.age.mean() # Lấy số trung bình của tuổi
print(df_age_mean)
```

```python
df.age.map(lambda age: age - df_age_mean)
```

```python
def remean_age(row):
    row.age = row.age - df_age_mean
    return row

df.apply(remean_age, axis='columns') # Axis có 2 giá trị: 0 (hoặc 'index') và 1 (hoặc 'columns')
```
