````python
import pandas as pd
````

````python
reviews = pd.read_csv('../test.csv')
reviews.head()
````

``` .rename()```    =   Để cho phép bạn thay đổi tên chỉ mục and/or tên cột.

````python
reviews.rename(columns={'Work/Study Hours': 'Sleep Hours'}) # Đổi tên cột
````

````python
reviews.rename(index={0: 'first person', 1: 'second person'}) # Đổi tên hàng
````

````python
# rename index thường ít dùng nên dùng set_index vì tiện hơn
reviews.set_index('id')
````

````python
# rows index và columns index đều có thể có thuộc tính name riêng. 
# Phương thức rename_axis() bổ sung có thể được sử dụng để thay đổi các tên
# axis : 0 là rows, 1 là columns
reviews.rename_axis('STT', axis=0).rename_axis('Tên cột', axis=1)
````

````python
# Ví dụ dễ nhìn cho đổi tên cột và hàng
df = pd.DataFrame([[1, 3], [2, 4]], columns=['A', 'B'])
df2 = df.rename_axis('STT', axis='rows').rename_axis('Tên cột', axis=1)
print(df2)
````
