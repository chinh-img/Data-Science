# Inner join

Giữ chỉ các dòng có khóa trùng nhau ở cả 2 bảng

```python
import pandas as pd
```

```python
table_1 = pd.DataFrame({
    "A" : ["A2", "A3", "A4"],
    "B" : ["B2", "B3", "B4"],
    "C" : ["C2", "C3", "C4"],
})

table_2 = pd.DataFrame({
    "C" : ["C1", "C2", "C4", "C5"],
    "D" : ["D1" ,"D2", "D4", "D5"],
})
```

```python
table_1
```

```python
table_2
```

```python
table_1 = table_1.merge(table_2, on="C", how="inner")
```

```python
table_1
```

# Left join

Giữ toàn bộ dòng bên trái, thêm dữ liệu từ bảng phải nếu có

```python
table_1 = pd.DataFrame({
    "A" : ["A2", "A3", "A4"],
    "B" : ["B2", "B3", "B4"],
    "C" : ["C2", "C3", "C4"],
})

table_2 = pd.DataFrame({
    "C" : ["C1", "C2", "C4", "C5"],
    "D" : ["D1" ,"D2", "D4", "D5"],
})

print(table_1)
```

```python
table_1 = table_1.merge(table_2, on="C", how="left")
```

```python
table_1
```

# Right join

Giữ toàn bộ dòng bên phải

```python
left_table = pd.DataFrame({
    "A" : ["A2", "A3", "A4"],
    "B" : ["B2", "B3", "B4"],
    "C" : ["C2", "C3", "C4"],
})

right_table = pd.DataFrame({
    "C" : ["C1", "C2", "C4", "C5"],
    "D" : ["D1" ,"D2", "D4", "D5"],
})

print(left_table)
print(right_table)
```

```python
left_table = left_table.merge(right_table, on="C", how="right")
```

```python
left_table
```

# Outer join

Giữ tất cả dòng, kể cả không trùng khóa (NaN nếu thiếu)

```python
left_table = pd.DataFrame({
    "A" : ["A2", "A3", "A4"],
    "B" : ["B2", "B3", "B4"],
    "C" : ["C2", "C3", "C4"],
})

right_table = pd.DataFrame({
    "C" : ["C1", "C2", "C4", "C5"],
    "D" : ["D1" ,"D2", "D4", "D5"],
})

print(left_table)
print(right_table)
```

```python
left_table = left_table.merge(right_table, on="C", how="outer")
```

```python
left_table
```

# Self join

```python
sequels = pd.DataFrame({
    "id" : [19995, 862, 863, 597, 24428],
    "title" : ["Avatar", "Toy Story", "Toy Story 2", "Titanic", "The Ave"],
    "sequel" : [None, 863, 10193, None, None],
})
```

```python
original_sequels = sequels.merge(sequels, left_on="sequel", right_on="id", suffixes=("_original", "sequel"))
```
