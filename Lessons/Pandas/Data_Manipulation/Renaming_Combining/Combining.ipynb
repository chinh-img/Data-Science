```python
import pandas as pd
```

Validate: Các loại merge, join, concat trong pandas
- One-to-One: Mỗi giá trị trong cột khóa của DataFrame này khớp với đúng một giá trị trong cột khóa của DataFrame kia.
- Many-to-One: Nhiều giá trị trong cột khóa của DataFrame này khớp với một giá trị duy nhất trong cột khóa của DataFrame kia.
- Many-to-Many: Nhiều giá trị trong cột khóa của DataFrame này khớp với nhiều giá trị trong cột khóa của DataFrame kia.


```python
# concat() = Với một danh sách các phần tử, hàm này sẽ gộp các phần tử đó lại với nhau dọc theo một trục.
# TODO: chỉ dùng được cho dữ liệu khác nhau nhưng có cùng trường (cột)
# Độ phức tạp: dễ
jp_videos = pd.read_csv('./JPvideos.csv', encoding='utf-8', encoding_errors='replace') # vì jp có nhiều kí tự đặc biệt nên cần encoding
us_videos = pd.read_csv('./USvideos.csv')

pd.concat([jp_videos, us_videos], axis=0, ignore_index=False, keys=['JP', 'US']) 
```

```python
# join() : kết hợp các đối tượng DataFrame khác nhau có chung một chỉ mục.
# Độ phức tạp: vừa 
left = jp_videos.set_index(['title', 'trending_date'])
right = us_videos.set_index(['title', 'trending_date']) # set_index(): đặt chỉ mục

left.join(right, lsuffix='_JP', rsuffix='_US', how='outer') # lsuffix, rsuffix: hậu tố thêm vào tên cột để phân biệt
# how='outer': giữ tất cả các hàng từ cả hai DataFrame, nếu không có giá trị tương ứng thì sẽ là NaN
```

```python
# merge() : kết hợp các đối tượng DataFrame khác nhau dựa trên các cột chung
# Độ phức tạp: vừa
# Ví dụ: kết hợp hai DataFrame dựa trên cột 'title' và 'trending_date'
merged_videos = pd.merge(left=jp_videos, right=us_videos, 
                         on=['title', 'trending_date'], # cột chung để merge
                         suffixes=('_JP', '_US'),
                         how='outer')
merged_videos
```

# Merge_ordered

Dùng khi:
- Ordered data/ time series
- Fill missing values

```python
aapl = pd.DataFrame({
    "date": ["2007-02-01", "2007-03-01", "2007-04-01", "2007-05-01", "2007-06-01"],
    "close": [12.087143, 13.272857, 14.257143, 17.312857, 17.434286]
})

mcd = pd.DataFrame({
    "date": ["2007-01-01", "2007-02-01", "2007-03-01", "2007-04-01", "2007-05-01"],
    "close": [44.349998, 43.689999, 45.049999, 48.279999, 50.549999]
})
```

```python
pd.merge_ordered(aapl, mcd, on="date", suffixes=('_aapl' ,'_mcd'), fill_method='ffill') # ffile (Forward fill)
```
