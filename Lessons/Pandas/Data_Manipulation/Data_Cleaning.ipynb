# Data cleaning 🧹

The process of fixing/ removing: incomplete, incorrect, or irrelevant (không liên quan) data.

~75% of work done with Pandas is data cleaning

```python
import pandas as pd
```

```python
df = pd.read_csv("./pokemon.csv")

df
```

## Drop irrelevant columns

```python
# 1. Drop irrelevant columns
df = df.drop(columns=["Legendary", "No"])
```

```python
df
```

## Handle missing data

```python
# df = df.dropna(subset=["Type 2"]) # dropna là viết tắc của drop not available
df = df.fillna({"Type 2": "None"})
```

```python
df
```

## Fix inconsistent values

```python
df["Type 1"] = df["Type 1"].replace({"Grass": "GRASS"})
```
