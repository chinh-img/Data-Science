````python
import pandas as pd
````

````python
# Reading csv file
df = pd.read_csv("../test.csv")
df.head()
````

### 1. Index-based selection (Chọn dựa trên chỉ số): iloc[row, col]

```iloc[row, col]``` : row và col là index trong dataframe

````python
# Select first row of dataframe
df.iloc[0] # iloc và loc giống nhau về syntax: [row, column] nhưng cách dùng khác nhau
````

````python
# Get a column of dataframe
df.iloc[:, 3] # lấy column đầu. Dấu : là lấy đến n row trong dataframe
````

````python
df.iloc[:3, :2]
````

````python
df.iloc[[0,1,2], 0] # = df.iloc[:3,0]
````

````python
# Select just second and third entries
df.iloc[1:3, 0]
````

````python
df.iloc[-5:] # số âm thì lấy từ cuối data đi lên
````

### 2. Label-based selection (Chọn dựa trên nhãn): loc[row, col]

```loc[row, col]``` : row là index nhưng col là data index value (VD: Gender)

````python
# Get the first entry
df.loc[0, 'Gender']
````

````python
df.loc[:, ['Name', 'Gender', 'Age']]
````
