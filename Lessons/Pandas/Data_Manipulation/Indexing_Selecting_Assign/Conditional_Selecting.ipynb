```python
import pandas as pd
```

```python
# Read csv file
df = pd.read_csv('../test.csv')
df.head()
```

### Manipulating the index

```python
df.set_index("id") # Chọn index là id
```

### Conditional selection

```python
df['Work/Study Hours'] == 0.0
```

```python
df.loc[df['Work/Study Hours'] == 0.0, "Work/Study Hours"]
```

```python
df.loc[(df['Work/Study Hours'] == 0.0) & (df.Age <= 30) & (df['Dietary Habits'] != 'Healthy')]
```

```python
# .isin() chọn dữ liệu có giá trị "nằm trong" một danh sách các giá trị.
df.loc[df.City.isin(['Rajkot', 'Kolkata'])]
```

```python
# isnull và notnull
df.loc[df['Academic Pressure'].notnull()]
```
