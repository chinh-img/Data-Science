```python
import pandas as pd
```

```python
# Reading csv file
df = pd.read_csv("../test.csv")
df.head()
```

### 1. Selecting one Column

#### 1.1 Syntax 1

```python
# Selecting a column with [] (prefered way to select a column)
df["Gender"]
```

```python
# Check out a data type of a column
type(df["Gender"])
```

```python
# Series: attributes and methods
df['Gender'].index
df['Gender'].head()
```

#### 1.1 Syntax 2

```python
# Selecting a column with .
df.Gender
```

```python
# Selecting a column with . (pitfalls)
# df.Work Pressure # -> ERROR
```

```python
# Selecting the same column using []
df['Work Pressure']
```

### 2. Selecting two or MORE column

```python
# Select 2 columns using [[]]
df[['Gender', 'Work Pressure']] # theo order đặt: Gender trước, Work Pressure sau
```

```python
# Check out the data type of selection
type(df[['Gender', 'Work Pressure']]) 
# => Khi dùng [[]] ta sẽ nhận về 1 DataFrame, còn [] thì nhận về Series (mảng một chiều)
```

```python
# Select 2 or more columns using [[]]
df[['Gender', 'Work Pressure', 'CGPA', 'Study Satisfaction','Job Satisfaction']]
```
