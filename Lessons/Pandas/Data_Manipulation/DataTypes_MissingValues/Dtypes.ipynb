```python
import pandas as pd
```

Kiểu dữ liệu cho một cột trong DataFrame hoặc Series được gọi là dtype .

```python
reviews = pd.read_csv('../titanic3.csv')
reviews.head()
```

```python
# thuộc tính dtype để lấy kiểu của một cột cụ thể
reviews.age.dtype
```

```python
# thuộc tính dtypes trả về dtype của mọi cột trong DataFrame
reviews.dtypes
```

- float64: Số thực với độ chính xác cao (64-bit)
- int64: số nguyên có độ chính xác cao (64-bit) và có kích thước tương tự int trong Python
- object: chỉ bao gồm chuỗi (string) không có kiểu cụ thể

```python
# astype(): đổi một cột thuộc kiểu này sang kiểu khác (gọi nhanh: ép kiểu)
reviews.survived.astype('float64')
```

```python
# Index của Dataframe hay Series cũng có dtype riêng
reviews.index.dtype
```
