```python
import pandas as pd
```

```python
reviews = pd.read_csv('../titanic3.csv')
reviews.head()
```

Các mục thiếu giá trị (missing values) sẽ được gán là NaN (Not a Number). Vì lí do kĩ thuật, các giá trị
NaN sẽ coi là float64

```python
# phương thức pd.isnull(): chọn các mục NaN
# phương thức pd.notnull(): ngược lại isnull()
reviews[pd.isnull(reviews.age)]
```

```python
# Thay thế các giá trị bị thiếu là một thao tác phổ biến. 
# Pandas cung cấp một phương thức rất tiện dụng cho vấn đề này: fillna()
reviews.age.fillna(0)
```

```python
# replace() : thay 1 giá trị
# Phương thức replace() đáng được đề cập ở đây vì nó tiện dụng để thay thế dữ liệu bị thiếu khi dữ liệu được gán một số giá trị canh gác trong tập dữ liệu
reviews.age.replace('Missing value', 0)
```
