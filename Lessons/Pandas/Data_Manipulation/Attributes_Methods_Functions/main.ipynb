**Attribute** = 1 value liên kết với 1 object (được tham chiếu bằng tên thông qua các biểu thức có chấm)

**Function** = một nhóm các câu liên quan thực hiện một nhiệm vụ cụ thể

**Method** = 1 function được định nghĩa bên trong body của class

```python
import pandas as pd
```

```python
# Reading the csv file
df = pd.read_csv('../titanic3.csv')
```

```python
# Showing the dataframe
df
```

### 1. Attributes

```python
# getting access to shape attribute
df.shape # return a tuple(nums_of_rows, nums_of_cols)
```

```python
# getting access to the index attribute
df.index
```

```python
# getting access to the column attribute
df.columns
```

```python
# Data types of each column
df.dtypes # object ở đây khá giống string (dtypes: data types)
```

### 2. Methods

```python
# show first 5 rows in a dataframe
df.head(5) # head(n), n = the number of rows to display (default n = 5)
```

```python
# Showing the info of the dataframe
df.info()
```

```python
# describing the basic statistics of the dataframe
df.describe()
# count = number of rows in each columns
# mean (số trung bình)
# std (standard deviation)
# 25%, 50%, 75% = Q1, Q2, Q3 (vùng miền)
```

### 3. Functions

```python
# Obtaning the length of the dataframe (number of rows)
len(df)
```

```python
# Obtaining the highest index of the dataframe
max(df.index)
```

```python
# Obtaining the lowest index of the dataframe
min(df.index)
```

```python
# Obtaining the data type
type(df)
```

```python
# Rounding the values of the dataset
round(df, 2)
```
