```python
import pandas as pd
df = pd.read_csv('../test.csv')
```

```python
df.head()
```

### 1 Grouping

```python
# groupby() : nhóm dữ liệu theo 1 cột hoặc nhiều cột
df.groupby('Age').Age.count() # giống df['Age'].value_counts()
```

```python
df.groupby('Work/Study Hours')['Work/Study Hours'].min()
```

```python
df.groupby('id').apply(lambda df: df['Profession'].iloc[0])
```

```python
df.groupby(['Age', 'Profession']).apply(lambda df: df.loc[df.Age.idxmax()]) # lấy dòng có tuổi lớn nhất trong mỗi nhóm
```

```python
# .agg() trong pandas dùng để áp dụng một hoặc nhiều hàm thống kê/tổng hợp (aggregation functions) 
# lên từng nhóm dữ liệu sau khi đã groupby
df.groupby(['City']).Age.agg([len, min, max])
```

### 2. Multi Indexces

```python
city_reviewed = df.groupby(['City', 'Profession']).Degree.agg([len])
city_reviewed
```

```python
mi = city_reviewed.index
type(mi)
```

```python
city_reviewed.reset_index() # chuyển index thành cột
```
