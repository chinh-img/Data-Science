#### 1.2 Adding a new column with an array

The dataframe has 93800 rows, so we need to create an array with 93800 elements

```python
import numpy as np
```

```python
# Create an array of 93800 elements
life_pressure = np.arange(0, 93800)
```

```python
# Lenght of the array
len(life_pressure)
```

```python
# Adding a new column
df['Life Pressure'] = life_pressure
df
```

```python
# Create a random integer from 1 to 100
int_life_pressure = np.random.randint(1, 100, size=93800)
```

```python
# Min value inclusive and high value exclusive
min(int_life_pressure)
max(int_life_pressure)
```

```python
# adding a new column
df['Life Pressure'] = int_life_pressure
df
```
