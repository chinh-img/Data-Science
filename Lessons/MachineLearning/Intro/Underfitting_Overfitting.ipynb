# Underfitting (quá đơn giản hóa mô hình)

- Là hiện tượng mô hình không học đủ từ dữ liệu huấn luyện → kết quả kém cả trên training và testing.
- Mô hình quá đơn giản (ví dụ: tuyến tính với dữ liệu phi tuyến tính) nên không nắm bắt được quy luật thực tế.

=> Khi một mô hình không nắm bắt được những điểm khác biệt và mô hình quan trọng trong dữ liệu, khiến nó hoạt động kém ngay cả trong dữ liệu huấn luyện

## 🧩 Biểu hiện:


1. Training error cao
2. Test error cũng cao
3. Mô hình có bias cao (high bias), variance thấp

📊 Ví dụ: Dùng linear regression để mô phỏng dữ liệu có dạng parabol.

## Overfitting (quá khớp mô hình)

- Là hiện tượng mô hình học quá kỹ dữ liệu huấn luyện, đến mức nhớ cả nhiễu (noise) → kết quả tốt trên training nhưng rất kém trên dữ liệu mới.
- Mô hình quá phức tạp, có quá nhiều tham số.

=> khi một mô hình khớp gần như hoàn hảo với dữ liệu huấn luyện, nhưng lại kém hiệu quả trong việc xác thực và xử lý dữ liệu mới khác

## 🧩 Biểu hiện:


1. Training error thấp
2. Test error cao
3. Mô hình có bias thấp, variance cao

📊 Ví dụ: Dùng polynomial regression bậc 10 để mô phỏng dữ liệu thực ra chỉ cần bậc 2.
