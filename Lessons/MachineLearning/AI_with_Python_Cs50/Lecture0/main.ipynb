# Search 🔎

Tìm kiếm giải pháp cho một vấn đề, như một ứng dụng định hướng tìm ra tuyến đường tốt nhất từ nơi bắt đầu đến đích, hoặc như chơi một trò chơi và tìm ra bước tiếp theo.

## Example 🤔

- 15 puzzle
- Maze (mê cung)
- Google map chỉ đường tiện nhất từ A đến B

## Agent 🤯 = Một thực thể nhận thức môi trường và hành động dựa trên môi trường đó

**```Ví dụ```**: trong một ứng dụng định hướng, thực thể này sẽ là biểu tượng của một chiếc xe cần quyết định các hành động để đến đích.

## State 🤖 = 1 số cấu hình của **agent** trong môi trường của nó

Initial state 1️⃣ = trạng thái agent bắt đầu. (Hiểu đơn giản là random puzzle mới khi chơi lại)

## Actions 👊 = những hành động **có thể** ở *bất kì* trạng thái nào

## Transition model 🔀 = mô tả nhận được sau khi thực hiện một số **action** trong 1 **state** khác nhau

## State space = tập hợp các state có thể nhận được từ initial state qua bất kỳ chuỗi các action nào
