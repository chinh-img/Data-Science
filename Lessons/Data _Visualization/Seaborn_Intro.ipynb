```python
import pandas as pd
pd.plotting.register_matplotlib_converters() # Cần thiết để tránh cảnh báo về định dạng ngày tháng trong matplotlib
import matplotlib.pyplot as plt
# Hiển thị đồ thị trong Jupyter Notebook
# %matplotlib inline để đảm bảo đồ thị hiển thị đúng, nếu không có nó, đồ thị có thể không hiển thị hoặc hiển thị không đúng cách trong notebook
%matplotlib inline
import seaborn as sns 
print("Setup Complete")
```

```python
# Path of the file to read
file_path = "./fifa.csv"

# Read the file into a variable fifa_data
data = pd.read_csv(file_path)
```

```python
data.head()
```

```python
# Set the width and height of the figure
plt.figure(figsize=(16,6), facecolor="gray") # plt.figure dùng để thiết lập kích thước của đồ thị (canvas)

# Line chart showing how FIFA rankings evolved over time 
sns.lineplot(data=data)

plt.show()
```
