```python
x = np.array([2023, 2024, 2025, 2026]) # Chọn Trục hoành
y = np.array([15, 25, 30, 20]) # Chọn Trục tung (Có thể điền string)

plt.plot(x, y) # Nếu chỉ truyền 1 args thì plt sẽ lấy dữ liệu đó làm trục tung
plt.show() # TODO:Phải có dòng này thì mới thấy được biểu đồ
```

# Plot customization 🎨

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])


line_style = dict(marker=".", # Biểu tượng các vị trí điểm x bằng marker (., o, v, *,... đọc doc)
                  markersize=30, # markersize có thể để là ms=30
                  markerfacecolor="#1cd3fc", # Có thể chỉnh color bằng tên, RGB, HEX, HSL (mfc)
                  markeredgecolor="#fc1c1c",
                  linestyle="solid", # dashed, dotted, dashdot, None, solid
                  linewidth=4,
                  )

plt.plot(x, y1, color="#23b06e",**line_style) 

plt.plot(x, y2, color="#dd00ff", **line_style)

plt.plot(x, y3, color="#cad102", **line_style)

plt.show()
```

# Grid Lines ⏹

```python
# grid() = Helps makes plots easier to read by avoiding reference lines

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis='y',
         linewidth=2,
         color="lightgray",
         linestyle="dashed")

plt.plot(x, y)
plt.show()
```

# Bar charts 📊

**Bar chart** = compare categories of data by representing each category with a bar

```python
categories = ["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweet"]
values = np.array([4, 3, 2, 5, 3, 1])

plt.bar(categories, values, color="skyblue")
# plt.barh(categories, values, color="skyblue") # Horizontal bar charts

plt.title("Daily consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

# Pie charts 🥧

**Pie charts** = Biểu đồ tròn được chia thành các lát để thể hiện tỷ lệ phần trăm của tổng số.

Tốt để trực quan hóa sự phân bổ giữa các danh mục.

```python
categories = ['Freshmen', "Sophomores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])
colors = ["red", "yellow", "blue", "green"]

plt.pie(values, labels=categories,
                autopct="%1.1f%%", # auto percentage
                colors=colors,
                explode=[0, 0, 0, 0.1],
                shadow=True,
                startangle=90)

plt.title("Sigma college")

plt.show()
```

# Scatter graph (biểu đồ phân tán)

- Mục đích: Hiển thị mối quan hệ giữa hai biến.
- Giúp xác định tương quan (tương quan dương +, tương quan âm -, hoặc không có tương quan).
- Ví dụ: Mối quan hệ giữa thời gian học và điểm kiểm tra

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) # Hours studied
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # Grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # Hours studied
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) # Grades

plt.scatter(x1, y1, color="red",
                    alpha=0.5, # Transparent
                    s = 200, # Size
                    label="Class A")

plt.scatter(x2, y2, color="green",
                    alpha=0.5, # Transparent
                    s = 200, # Size
                    label="Class B")

plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")

plt.legend() # Gọi chú thích
plt.show()
```

# Histograms

- Biểu diễn trực quan về sự phân bố của dữ liệu định lượng.
- Họ nhóm các giá trị vào các ngăn (khoảng) và đếm số lượng giá trị nằm trong mỗi phạm vi.

```python
scores = np.random.normal(loc=80, scale=10, size=100) # scale là độ lệch chuẩn quanh trung vị loc
scores = np.clip(scores, 0, 100) # Set min và max boundaries

plt.hist(scores, bins=10,
                color="lightgreen",
                edgecolor="black",)

plt.title("Exam Scores")
plt.xlabel("Scores")
plt.ylabel("# of students")

plt.show() 
```

# Subplot

- Figure = The entire canvas 🖼️
- Ax = A single plot (subplot) 📈

```python
print(plt.subplots(2, 2)) # Trả về tuple: Figure (canva của chúng ta), và axes: mảng 2 chiều của numpy
```

```python
x = np.array([1, 2, 3,4 , 5])

figure, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, x*2, color="red")
axes[0, 0].set_title("x*2")

axes[0, 1].bar(x, x**2, color="blue")
axes[0, 1].set_title("x**2")

axes[1, 0].scatter(x, x**3, color="green", alpha=0.5, s=100)
axes[1, 0].set_title("x**3")

axes[1, 1].plot(x, x**4, color="purple")
axes[1, 1].set_title("x**4")

plt.tight_layout() # Tránh subplot chồng lên nhau
plt.show()
```
